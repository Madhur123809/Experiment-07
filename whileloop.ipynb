```python
# Reverse a number
num = int(input("Enter a Number:"))
rev = 0
while num > 0:
     digit = num %10
     rev = rev * 10 + digit
     num //=10
print("Reversed number =",rev)
```

```python
#check palindrome number/string
num = int(input("Enter a number :"))
temp = num
rev = 0
while num>0:
  rev = rev * 10 + num % 10
  num //=10
if temp == rev:
   print("Palindrome")
else :
  print("NOT Palindrome")
```

```python
s = "madam"
i, j = 0, len(s) - 1
is_palindrome = True

while i < j:
    if s[i] != s[j]:
        is_palindrome = False
        break
    i += 1
    j -= 1

if is_palindrome:
    print("Yes")
else:
    print("No")
```

```python
#check Palindrome number
st = input("Enter the string :")
rev = st[::-1]
if st==rev:
  print("Palindrome")
else:
  print("not palindrome")
```

```python
#count digits in a number
num=int(input("Enter a num:"))
count = 0
while num> 0 :
      count+=1
      num = num//10
print("numberof digits=",count)
```

```python
#Search an elemnent in a list
nums = [10,20,30,40,50]
key = int(input("Enter element to search :"))

i = 0
while i <len(nums):
   if nums[i]==key:
        print("element found at key:",i)
        break
   i+=1
else:
  print("element not found")
```
